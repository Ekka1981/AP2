````
import pandas as pd

# Map contract components to their base values and GST rates from df_pricing
# Supply: Plant & Equipment (S.No 1.1)
base_plant = df_pricing.loc[df_pricing['S.No.'] == '1.1', 'Base Price Excl. GST (INR)'].values[0]
gst_rate_plant = df_pricing.loc[df_pricing['S.No.'] == '1.1', 'Applicable GST %'].values[0]

# Supply: Mandatory Spares (S.No 1.2)
base_spares = df_pricing.loc[df_pricing['S.No.'] == '1.2', 'Base Price Excl. GST (INR)'].values[0]
gst_rate_spares = df_pricing.loc[df_pricing['S.No.'] == '1.2', 'Applicable GST %'].values[0]

# Services: Erection & Testing (S.No 2.2)
base_erection = df_pricing.loc[df_pricing['S.No.'] == '2.2', 'Base Price Excl. GST (INR)'].values[0]
gst_rate_erection = df_pricing.loc[df_pricing['S.No.'] == '2.2', 'Applicable GST %'].values[0]

# Services: Civil & Allied Works (S.No 2.3)
base_civil = df_pricing.loc[df_pricing['S.No.'] == '2.3', 'Base Price Excl. GST (INR)'].values[0]
gst_rate_civil = df_pricing.loc[df_pricing['S.No.'] == '2.3', 'Applicable GST %'].values[0]

# Services: Freight & Insurance (S.No 2.1)
base_freight = df_pricing.loc[df_pricing['S.No.'] == '2.1', 'Base Price Excl. GST (INR)'].values[0]
gst_rate_freight = df_pricing.loc[df_pricing['S.No.'] == '2.1', 'Applicable GST %'].values[0]

# Map each component to its base value and GST rate
component_mapping = {
    'Supply: Plant & Equipment': (base_plant, gst_rate_plant),
    'Supply: Mandatory Spares': (base_spares, gst_rate_spares),
    'Services: Erection & Testing': (base_erection, gst_rate_erection),
    'Services: Civil & Allied Works': (base_civil, gst_rate_civil),
    'Services: Freight & Insurance': (base_freight, gst_rate_freight)
}

# Populate Base Value, Option 1 Payout Excl. GST, and Option 1 Payout Incl. GST
for idx, row in df_milestones.iterrows():
    comp = row['Contract & Scope Component']
    opt1_pct = row['Option 1 % (With Advance)']

    if comp in component_mapping:
        base_val, gst_rate = component_mapping[comp]
        df_milestones.at[idx, 'Base Value Excl. GST (INR)'] = base_val
        if pd.notna(opt1_pct):
            payout_excl = base_val * opt1_pct
            payout_incl = payout_excl * (1 + gst_rate)
            df_milestones.at[idx, 'Option 1 Payout Excl. GST (INR)'] = payout_excl
            df_milestones.at[idx, 'Option 1 Payout Incl. GST (INR)'] = payout_incl

# Calculate Subtotals and Totals for Option 1 Payouts
sub_1a_mask = df_milestones['Contract & Scope Component'] == 'Supply: Plant & Equipment'
df_milestones.loc[df_milestones['Clause Ref'] == 'Sub-1A', 'Option 1 Payout Excl. GST (INR)'] = df_milestones.loc[sub_1a_mask, 'Option 1 Payout Excl. GST (INR)'].sum()
df_milestones.loc[df_milestones['Clause Ref'] == 'Sub-1A', 'Option 1 Payout Incl. GST (INR)'] = df_milestones.loc[sub_1a_mask, 'Option 1 Payout Incl. GST (INR)'].sum()

sub_1b_mask = df_milestones['Contract & Scope Component'] == 'Supply: Mandatory Spares'
df_milestones.loc[df_milestones['Clause Ref'] == 'Sub-1B', 'Option 1 Payout Excl. GST (INR)'] = df_milestones.loc[sub_1b_mask, 'Option 1 Payout Excl. GST (INR)'].sum()
df_milestones.loc[df_milestones['Clause Ref'] == 'Sub-1B', 'Option 1 Payout Incl. GST (INR)'] = df_milestones.loc[sub_1b_mask, 'Option 1 Payout Incl. GST (INR)'].sum()

sub_2_mask = df_milestones['Contract & Scope Component'].str.startswith('Services:')
df_milestones.loc[df_milestones['Clause Ref'] == 'Sub-2', 'Option 1 Payout Excl. GST (INR)'] = df_milestones.loc[sub_2_mask, 'Option 1 Payout Excl. GST (INR)'].sum()
df_milestones.loc[df_milestones['Clause Ref'] == 'Sub-2', 'Option 1 Payout Incl. GST (INR)'] = df_milestones.loc[sub_2_mask, 'Option 1 Payout Incl. GST (INR)'].sum()

# Total EPC
epc_excl = df_milestones.loc[df_milestones['Clause Ref'].isin(['Sub-1A', 'Sub-1B', 'Sub-2']), 'Option 1 Payout Excl. GST (INR)'].sum()
epc_incl = df_milestones.loc[df_milestones['Clause Ref'].isin(['Sub-1A', 'Sub-1B', 'Sub-2']), 'Option 1 Payout Incl. GST (INR)'].sum()
df_milestones.loc[df_milestones['Clause Ref'] == 'TOTAL EPC', 'Option 1 Payout Excl. GST (INR)'] = epc_excl
df_milestones.loc[df_milestones['Clause Ref'] == 'TOTAL EPC', 'Option 1 Payout Incl. GST (INR)'] = epc_incl

# Display calculated DataFrame
pd.options.display.float_format = '{:,.2f}'.format
display(df_milestones[['Clause Ref', 'Contract & Scope Component', 'Option 1 % (With Advance)', 'Option 1 Payout Excl. GST (INR)', 'Option 1 Payout Incl. GST (INR)']])
````

Based on the master commercial model and contract terms for the **234 MW (AC) Maithon Floating Solar PV Project**, the payment milestones and advance recovery terms are structured into distinct schedules across the Supply, Services, and O&M contracts.

---

### **1. Mobilization Advance & Security Terms**

* **Advance Quantum:** A **10% interest-free mobilization advance** is available under the First Contract (Supply of Plant & Equipment) upon contract signing, submission of an **110% Advance Bank Guarantee (ABG)**, a **3% Contract Performance Bank Guarantee (CPBG)**, and an approved L2 Work Schedule.
* **Interest & Penalties:** While the advance is interest-free during regular progress, any delay in milestone achievement or payment recovery can trigger delay interest calculated at the **SBI Base Rate + 3.50% p.a. (total ~12.25% p.a.)** under Clause 5.0.

---

### **2. First Contract: Supply of Plant & Equipment (Milestones & Recovery)**

Under **Option 1 (With Advance)**, progressive dispatch and delivery milestones are structured as follows:

1. **Advance (10%):** Paid against contract signing and 110% ABG submission.
2. **Dispatch Payout (55%):** Released pro-rata upon equipment dispatch based on the Billing Break-Up (BBU), backed by invoices, Railway Receipts/Lorry Receipts (LR/RR), and Material Dispatch Clearance Certificates (MDCC).
3. **Site Receipt (15%):** Released upon physical arrival at site, verification, and issuance of the **Material Receipt Certificate (MRC)**.
4. **Erection & Testing (5%):** Released against successful erection, testing, and commissioning on a pro-rata basis.
5. **Operational Acceptance (5%):** Released upon successful Performance Ratio (PR) guarantee tests and submission of O&M manuals.
6. **Final Retention Holdback (10%):** Released upon facility completion and successful demonstration of the **1st Year Annual CUF Guarantee** (or released early by deploying an 18-month Retention Replacement Bank Guarantee).

---

### **3. Pro-Rata Advance Recovery Terms**

* **Proportional Deduction:** As tracked in **Tab 9 (`9_RA_Bill_&_Advance_Tracker`)**, the 10% mobilization advance is recovered **pro-rata (at 10% of the gross dispatch value)** across progressive equipment dispatches.
* **Exhaustion Schedule:** The deduction is applied directly against the 55% dispatch billable payouts until the advance balance is **100% fully recovered** exactly as equipment deliveries conclude across the 18-month execution window.

---

### **4. Second Contract: Services (Erection, Civil Works & Freight)**

* **Erection & Testing:** 80% paid progressively on equipment installation and quality clearance; 10% upon Operational Acceptance; and **10% final retention** released after 1st-year CUF demonstration (or via 18M BG).
* **Civil & Allied Works:** 80% progressively on milestone completion; 10% upon Operational Acceptance; and **10% final retention** released post 1st-year CUF verification.
* **Freight & Insurance:** 100% paid pro-rata to the value of equipment/spares received at site upon invoice submission.

---

### **5. Third Contract: Comprehensive O&M (LOA 3091)**

* Paid in **quarterly installments** across the 10-year operating horizon, subject to annual generation guarantees and performance ratio compliance under Clause 8.0.

* **Upfront Liquidity Benefit:** **INR 994.51 Million** (disbursed interest-free against a 110% Advance Bank Guarantee).
* **Interest Savings (6-Month Base Case):**
  * **At 8.75% SBI Base Rate:** **INR 43.51 Million** in financing costs saved.
  * **At 10.50% Alternative Rate:** **INR 52.21 Million** in financing costs saved.

#### **4. Sensitivity Insights**
* **Holding Period:** Every month of delaying the equipment dispatch stage increases nominal working capital interest savings by **INR 7.25 Million** (at 8.75% cost of capital) to **INR 8.70 Million** (at 10.50% cost of capital).
* **Cost of Capital:** For every **1% shift** in your organization's borrowing rate, the cash utility of securing Option 1 shifts linearly by **INR 4.97 Million** over a 6-month holding window.

````
import numpy as np
import pandas as pd

# Define the advance amount including GST
advance_amount_val = 994510266.55
time_period_yrs = 0.5  # 6 Months

# Define cost of capital rates from 5% to 15% with 1% increments
sensitivity_rates = np.arange(0.05, 0.16, 0.01)

# Calculate interest savings for each rate
sensitivity_records = []
for r in sensitivity_rates:
    interest_savings = advance_amount_val * r * time_period_yrs
    monthly_savings = interest_savings / 6.0
    sensitivity_records.append({
        'Cost of Capital Rate (%)': f'{r*100:.1f}%',
        'Total Interest Savings (INR)': interest_savings,
        'Monthly Interest Savings (INR)': monthly_savings
    })

# Create DataFrame
df_sensitivity_5_15 = pd.DataFrame(sensitivity_records)

print("--- WORKING CAPITAL INTEREST SAVINGS SENSITIVITY (5% TO 15% COST OF CAPITAL) ---")
display(df_sensitivity_5_15.style.format({
    'Total Interest Savings (INR)': 'INR {:,.2f}',
    'Monthly Interest Savings (INR)': 'INR {:,.2f}'
}))
````
